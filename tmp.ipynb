```python
df = openpyxl.load_workbook('/Users/akeem/Downloads/Book2.xlsm')
sheets = df.sheetnames
ws = df[sheets[0]]

found = {}

# Remove column 2
ws.delete_cols(2)
saved_name = -1
#loop through rows, grab value in col 1. Check if it is a url
for row in ws.iter_rows(min_row=2, max_col=1):
    for cell in row:
        value = cell.value
        if isinstance(value, str) and value.startswith('http'):
            print(f"Found name: {saved_name} with url: {value}")
            found[saved_name] = value
            saved_name = -1
        elif isinstance(value, str) and ", " in value and "Also" not in value and "Story-" not in value:
            saved_name = value
            found[saved_name] = None
            #print(f"Found name: {saved_name}")

found
        
```

```python
found_df = pd.DataFrame.from_dict(found, orient='index', columns=['URL'])
found_df.reset_index(inplace=True)
found_df = found_df.rename(columns={'index': 'Name'})
found_df.to_csv('found_names_urls.csv', index=False)
```

```python
name_url_df = pd.read_csv("/Users/akeem/Downloads/found_names_urls.csv")
name_url_df["Count_Assigned"] = 0
#Add a column "Forum Title" to name_url_df
# and add a sequental value of VIS_001, VIS_002, etc. to the column
name_url_df["Forum Title"] = ["VIS_{:03d}".format(i) for i in range(1, len(name_url_df) + 1)]

name_url_df
```

```python
# For each reviewer name in the name column, randomly select 4 URLs from the URL column that are not associated with that name
# create a new dataframe with columns: Reviewer Name, Submission Name, URL and insert the data accordingly
# Add a column to the name_url_df called "count" that shows how many times each URL has been assigned
import random
final_df = pd.DataFrame(columns=["Reviewer Name", "Submitter Name", "URL", "Forum for URL"])
all_reviewers = name_url_df["Name"].tolist()
random.shuffle(all_reviewers)
reviews_per_reviewer = 4
reviewer_count_dict = {name: 0 for name in all_reviewers}
position = 0
for index, row in name_url_df.iterrows():
    if position + reviews_per_reviewer < len(all_reviewers):
        reviewers = all_reviewers[position:position + reviews_per_reviewer]
    else:
        from_end = len(all_reviewers) - position
        from_start = reviews_per_reviewer - from_end
        reviewers = all_reviewers[position:] + all_reviewers[:from_start]
    position = (position + reviews_per_reviewer) % len(all_reviewers)
    for reviewer in reviewers:
        final_df = pd.concat([final_df, pd.DataFrame({"Reviewer Name": [reviewer], "Submitter Name": [row["Name"]], "URL": [row["URL"]], "Forum for URL": [row["Forum Title"]]} )], ignore_index=True)
        name_url_df.loc[name_url_df["Name"] == row["Name"], "Count_Assigned"] += 1
        reviewer_count_dict[reviewer] += 1

    final_df.to_csv("/Users/akeem/Downloads/2210_asn.csv", index=False)
final_df
```
