## Git and GitHub Basics

We are using a remote repository on GitHub to manage our code. GitHub, or something equivalent, is used almost everywhere that code is developed. It serves several purposes, such as:
<ul>
<li> Version control - it allows us to keep track of changes to our code, and to revert to previous versions if we mess something up.</li>
<li> Collaboration - it allows multiple people to work on the same codebase and manage edits from different people. </li>
<li> Backup - it allows us to have a backup of our code in case something happens to our local copy. </li>
<li> Constant Integration and Testing - it allows us to have automated tests that run whenever we make changes to our code, to make sure that we don't break anything, or even have packaged versions build for distribution.</li>
</ul>

Using GitHub is a bit of a headache at first, but it is useful and worth learning. 

### Repositories

GitHub is organized around repositories, basically each 'thing' is a repository, and each repository can have multiple files in it. We have one repository for the workbooks in this class as well as one for each assignment you do. In reality, each program/project is normally its own repository. 

The main repository is held on GitHub's servers, and that is the "published" version of the code. Each person who works on the project will clone, or copy, that repository to their local machine, and then they can make changes to the code on their local copy. This is what you have all done with this repository - there is a central version online, and each of you have made your own copies to work on. When you 'pull' you are asking GitHub to get the most recent version of the code from the central repository and update your local copy with it. 

When working on code, people commit or push changes from their local machine to the central repository. For example, if I make a change to one of these workbooks on my local machine, I can then commit that change to the central repository, which will update the online version of the code. Each of you can then pull that change to get the most recent version of the code on your local machine. With teams of people working on large programs, this allows each person to independently work on their own local copy, and then share their changes with the rest of the team by pushing to the central repository. GitHub also manages conflicts - if two people make changes to the same part of the code at the same time, GitHub will flag this and allow the team to resolve the conflict before the changes are merged.

In addition, the central GitHub server can be setup to do things for us automatically, such as run tests, build projects, or publish results. This is called continuous integration, and it is a common practice in software development. For example, if we are creating a piece of code that does translations from one language to another, we can build some test scripts that feed our code several sample sentences, gets the responses, and checks them for accuracy. This type of setup allows for automated testing, so as people make changes or improve other parts of the code, there aren't accidental breakages introduced. If there is an error, this makes it easy to spot and roll back to a previous version if needed. This also leads to the idea of test driven development, where you aim to write tests that define what success is for your code is, then write code to meet those tests. 

Lastly, GitHub also allows for branching, or creating a copy of a project and customizing it on your own, without affecting the main version. This is commonly used in real software development where things are being changed or added to software - for example, if we are building a piece of software and we want to add a new feature, we can create a branch to work on that feature, and then once it is ready and tested, we can merge it back into the main version. This allows for development to happen in parallel without affecting the stability of the main codebase. Our usage of branches will be more limited, but each of you will be generating your own branch for each assignment, and if you try to push this main workbook repository up to your own GitHub profile.

#### GitHub Issues

Here are some common issues that have come up in the past working with GitHub:

<ul>
<li> I can't push my code to GitHub, I get an error about size or timeout. </li>
    <ul>
    <li> This is likely because you have a large file in your repository that is too big for GitHub to handle. Don't have files larger than 50MB in your repository.</li>
    <li> Running a command of "git config http.postBuffer 524288000" in a terminal, doing the push, then running "git config --unset http.postBuffer" to reset it back to normal, can be a workaround if you have a large file that you need to push. This changes the size of the buffer that Git uses to push data, so it can handle larger files, but it isn't assured to work. </li>
    </ul>
</ul>
